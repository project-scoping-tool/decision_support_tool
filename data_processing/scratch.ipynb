```python
# Read data from files (Will take long!)

excel_file_name = '5 YR 2019-2023 STATEWIDE CRASH DATA.xlsx'

print('Reading Excel files...')
data_crashes  = pl.read_excel(excel_file_name, sheet_name='2019-2023 STATEWIDE', read_options={'schema_sample_rows': None})
print('Read crashes sheet')
data_persons  = pl.read_excel(excel_file_name, sheet_name='PERSONS INVOLVED', read_options={'schema_sample_rows': None})
print('Read persons sheet')
```
